```python
import sqlite3

conn=sqlite3.connect("ClientesVarios.db")
cur= conn.cursor()
```

```python
#Lectura archivo CSV 1
import os
import csv

with open("Customers1.csv") as f:
    reader= csv.reader(f)
    data=list(reader)
```

```python
len(data)
```

```python
#Creacion de la tabla donde se importará el archivo csv
cur.execute("""
CREATE TABLE Customers1(
CustomerName TEXT PRIMARY KEY,
Address TEXT,
City TEXT,
Country TEXT
);
"""
)
```

```python
for renglon in range(1, len(data)):
    cur.execute("""
        INSERT INTO Customers1(CustomerName, Address, City, Country)
        VALUES(?,?,?,?)
    """, data[renglon])
```

```python
conn.commit()
```

```python
#Lectura archivo CSV 2
with open("Customers2.csv") as f:
    reader= csv.reader(f)
    data=list(reader)
```

```python
len(data)
```

```python
#Creacion de la tabla donde se importará el archivo csv
cur.execute("""
CREATE TABLE Customers2(
CustomerName TEXT PRIMARY KEY,
ContactName TEXT,
PostalCode TEXT,
PurchaseUSD INTEGER
);
"""
)
```

```python
for renglon in range(1, len(data)):
    cur.execute("""
        INSERT INTO Customers2(CustomerName, ContactName, PostalCode, PurchaseUSD)
        VALUES(?,?,?,?)
    """, data[renglon])
```

```python
conn.commit()
```

```python
#Revision de tabla 1
cur.execute("SELECT * FROM Customers1 LIMIT 10;")
result=cur.fetchall()
result
```

```python
#Revision de tabla 2
cur.execute("SELECT * FROM Customers2 LIMIT 10;")
result=cur.fetchall()
result
```

```python
#Consulta mediante Inner Join
cur.execute("""
SELECT p1.CustomerName, p1.City, p2.ContactName,p2.PurchaseUSD
FROM Customers1 AS p1 INNER JOIN Customers2 AS p2
ON p1.CustomerName= p2.CustomerName
""")
result= cur.fetchall()
result
```

```python
#Otra opción
cur.execute("""
SELECT p1.CustomerName, p1.City, ContactName, PurchaseUSD
FROM Customers1 AS p1 INNER JOIN Customers2 AS p2
ON p1.CustomerName= p2.CustomerName
""")
result= cur.fetchall()
result
```

```python
#Otra alternativa mediante el comando USING, solo cuando ambas tablas tinen el mismo nombre para el campo en comun
cur.execute("""
SELECT p1.CustomerName, p1.City, ContactName, PurchaseUSD
FROM Customers1 AS p1 INNER JOIN Customers2 as p2
USING(CustomerName) 
""")
result=cur.fetchall()
result
```

```python
#Consulta mediante LEFT JOIN
cur.execute("""
   SELECT p1.CustomerName, p1.City, p2.ContactName, p2.PurchaseUSD
   FROM Customers1 AS p1 LEFT JOIN Customers2 AS p2
   ON p1.CustomerName = p2.CustomerName
""")
result=cur.fetchall()
result
```

```python
#Uso de CASE WHEN para separar por clases los grupo de compra
cur.execute("""
   SELECT CustomerName,ContactName,
   CASE WHEN PurchaseUSD < 10000 Then "Venta Baja"
   WHEN PurchaseUSD <= 15000 Then "Venta Media"
   ELSE "Venta Alta" END AS Grupo_Venta
   FROM Customers2
   ORDER BY Grupo_Venta
""")
result=cur.fetchall()
result
```

```python
with open("Customers3.csv") as f:
    reader= csv.reader(f)
    data=list(reader)
```

```python
cur.execute("DROP TABLE IF EXISTS Customers3")

cur.execute("""
CREATE TABLE Customers3(
    CustomerName TEXT PRIMARY KEY,
    ContactName TEXT,
    PostalCode TEXT,
    PurchaseUSD INTEGER
);
""")

for renglon in range(1, len(data)):
    cur.execute("""
        INSERT INTO Customers3(CustomerName, ContactName, PostalCode, PurchaseUSD)
        VALUES(?,?,?,?)
    """, data[renglon])

conn.commit()
```

```python
#Uso de la operacion UNION
cur.execute("""
SELECT CustomerName,ContactName,PostalCode,PurchaseUSD
FROM Customers2
UNION
SELECT CustomerName,ContactName,PostalCode,PurchaseUSD
FROM Customers3
ORDER BY CustomerName
""")
result=cur.fetchall()
result
```

```python
#Uso de la operacion UNION ALL
cur.execute("""
SELECT CustomerName,ContactName,PostalCode,PurchaseUSD
FROM Customers2
UNION ALL
SELECT CustomerName,ContactName,PostalCode,PurchaseUSD
FROM Customers3
ORDER BY CustomerName
""")
result=cur.fetchall()
result
```

```python
#Uso de la operacion INTERSECT
cur.execute("""
   SELECT CustomerName,ContactName ,PostalCode,PurchaseUSD
   FROM Customers2
   INTERSECT
   SELECT CustomerName,ContactName ,PostalCode,PurchaseUSD
   FROM Customers3
   ORDER BY CustomerName
""")
result=cur.fetchall()
result
```

```python
#Uso de la operacion EXCEPT
cur.execute("""
   SELECT CustomerName,ContactName ,PostalCode,PurchaseUSD
   FROM Customers2
   EXCEPT
   SELECT CustomerName,ContactName ,PostalCode,PurchaseUSD
   FROM Customers3
   ORDER BY CustomerName
""")
result=cur.fetchall()
result
```

```python
#Ejemplo de la Semi-Join
cur.execute("""
SELECT CustomerName, Country FROM Customers1
WHERE CustomerName IN (SELECT CustomerName FROM Customers2 WHERE PurchaseUSD < 10000)
""")
result = cur.fetchall()
result
```

```python
#Ejemplo de AntiJoin
cur.execute("""
SELECT CustomerName, Country FROM Customers1
WHERE CustomerName NOT IN (SELECT CustomerName FROM Customers2 WHERE PurchaseUSD >= 10000)
""")
result = cur.fetchall()
result
```

```python
#Uso de Subquery para seleccionar aquellos clientes cuyo monto de compra es menor al promedio general
cur.execute("""
SELECT CustomerName, PurchaseUSD FROM Customers2
WHERE PurchaseUSD < ( SELECT AVG (PurchaseUSD) FROM Customers2)
""")
result = cur.fetchall()
result
```
