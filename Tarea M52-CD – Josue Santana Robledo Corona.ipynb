```python
import pandas as pd
import numpy as np
import warnings
warnings.filterwarnings("ignore")
```

```python
df = pd.read_csv('kc_house_data.csv')
print(df.shape)
df.head()
```

```python
Y = df['price'].values.reshape(-1, 1)
Y
```

```python
X_data = df.drop(columns=['id', 'date', 'zipcode', 'price',
                          'sqft_above', 'sqft_basement'])

# Agregar columna de 1's (intercepto)
unos = np.ones((X_data.shape[0], 1))
X = np.hstack([unos, X_data.values])

# Nombres de las columnas
nombres = ['const'] + X_data.columns.tolist()

print("Dimensiones X:", X.shape)
print("Nombres:", nombres)
X
```

```python
XT_X = np.matmul(np.matrix.transpose(X), X)
XT_X
```

```python
XT_X_inv = np.linalg.inv(XT_X)
XT_X_inv
```

```python
XT_Y = np.matmul(np.matrix.transpose(X), Y)
XT_Y
```

```python
betas = np.matmul(XT_X_inv, XT_Y)
betas
```

```python
Y_pred = np.matmul(X, betas)
Y_pred
```

```python
Resid = Y - Y_pred
Resid
```

```python
RSS = float(np.matmul(np.matrix.transpose(Resid), Resid))
RSS
```

```python
TSS = float(np.matmul(np.matrix.transpose(Y), Y) - len(Y) * (Y.mean()**2))
TSS
```

```python
R_cuad = float(1 - RSS / TSS)
R_cuad
```

```python
RSqAj = float(1 - (RSS / (X.shape[0] - X.shape[1])) / (TSS / (X.shape[0] - 1)))
RSqAj
```

```python
s_cuad = RSS / (len(Y) - X.shape[1])
s_cuad
```

```python
import math
s = math.sqrt(s_cuad)
s
```

```python
result_t = []
for i in range(0, X.shape[1]):
    t = float(betas[i] / (s * math.sqrt(XT_X_inv[i][i])))
    result_t.append(t)
result_t
```

```python
import scipy.stats
grados_libertad = len(Y) - X.shape[1]
t_critico = float(abs(scipy.stats.t.ppf(q=0.025, df=grados_libertad)))
t_critico
```

```python
#Criterio 1
for i in range(0, X.shape[1]):
    if abs(result_t[i]) > t_critico:
        print(f"Beta {i} ({nombres[i]}) SÍ es significativa")
    else:
        print(f"Beta {i} ({nombres[i]}) NO es significativa")
```

```python
# Criterio 2
for i in range(0, X.shape[1]):
    p = scipy.stats.t.sf(abs(result_t[i]), df=grados_libertad) * 2
    print(f"Valor p de Beta {i} ({nombres[i]}) = {p:.4e}")
```

```python
# Criterio 3
for i in range(0, X.shape[1]):
    li = float(betas[i]) - t_critico * s * math.sqrt(XT_X_inv[i][i])
    ls = float(betas[i]) + t_critico * s * math.sqrt(XT_X_inv[i][i])
    print(f"El valor de Beta {i} ({nombres[i]}) está entre {li:.4f} y {ls:.4f}")
```

### ENFOQUE 1: StepWise Forward

```python
def stepwise_forward(X, Y, nombres, alfa=0.01):
    n, k_total = X.shape
    incluidas = [0]  # intercepto
    candidatas = list(range(1, k_total))
    
    while True:
        mejor_p = 1.0
        mejor_var = None
        for j in candidatas:
            cols = incluidas + [j]
            X_test = X[:, cols]
            try:
                XT_X = np.matmul(X_test.T, X_test)
                XT_X_inv = np.linalg.inv(XT_X)
                XT_Y = np.matmul(X_test.T, Y)
                betas = np.matmul(XT_X_inv, XT_Y)
                Y_pred = np.matmul(X_test, betas)
                Resid = Y - Y_pred
                RSS = float(np.matmul(Resid.T, Resid))
                gl = n - len(cols)
                s = math.sqrt(RSS / gl)
                idx = len(cols) - 1
                t = float(betas[idx] / (s * math.sqrt(XT_X_inv[idx][idx])))
                p = scipy.stats.t.sf(abs(t), df=gl) * 2
                if p < mejor_p:
                    mejor_p = p
                    mejor_var = j
            except np.linalg.LinAlgError:
                continue
        if mejor_var is not None and mejor_p < alfa:
            incluidas.append(mejor_var)
            candidatas.remove(mejor_var)
        else:
            break
    return incluidas

cols_fwd = stepwise_forward(X, Y, nombres)
nombres_fwd = [nombres[i] for i in cols_fwd]
X_fwd = X[:, cols_fwd]

XT_X_f = np.matmul(X_fwd.T, X_fwd)
XT_X_inv_f = np.linalg.inv(XT_X_f)
XT_Y_f = np.matmul(X_fwd.T, Y)
betas_f = np.matmul(XT_X_inv_f, XT_Y_f)
Y_pred_f = np.matmul(X_fwd, betas_f)
Resid_f = Y - Y_pred_f
RSS_f = float(np.matmul(Resid_f.T, Resid_f))
TSS_f = TSS
R_cuad_f = float(1 - RSS_f / TSS_f)
RSqAj_f = float(1 - (RSS_f / (X_fwd.shape[0] - X_fwd.shape[1])) / (TSS_f / (X_fwd.shape[0] - 1)))
s_cuad_f = RSS_f / (len(Y) - X_fwd.shape[1])
s_f = math.sqrt(s_cuad_f)

result_t_f = []
for i in range(0, X_fwd.shape[1]):
    t = float(betas_f[i] / (s_f * math.sqrt(XT_X_inv_f[i][i])))
    result_t_f.append(t)

gl_f = len(Y) - X_fwd.shape[1]

print(f"N° parámetros: {X_fwd.shape[1]}")
print(f"R²        = {R_cuad_f:.6f}")
print(f"R² aj.    = {RSqAj_f:.6f}")
print(f"RSS       = {RSS_f:.4e}")
print(f"s         = {s_f:.4e}")
print()
for i in range(0, X_fwd.shape[1]):
    p = scipy.stats.t.sf(abs(result_t_f[i]), df=gl_f) * 2
    sig = "SÍ" if p < 0.05 else "NO"
    print(f"Beta {i} ({nombres_fwd[i]}): t={result_t_f[i]:.4f}, p={p:.4e} → {sig} significativa")
```

### Enfoque 2: StepWise Backward

```python
def stepwise_backward(X, Y, nombres, alfa=0.01):
    n, k = X.shape
    incluidas = list(range(k))
    
    while len(incluidas) > 1:
        cols = incluidas
        X_test = X[:, cols]
        try:
            XT_X = np.matmul(X_test.T, X_test)
            XT_X_inv = np.linalg.inv(XT_X)
            XT_Y = np.matmul(X_test.T, Y)
            betas = np.matmul(XT_X_inv, XT_Y)
            Y_pred = np.matmul(X_test, betas)
            Resid = Y - Y_pred
            RSS = float(np.matmul(Resid.T, Resid))
            gl = n - len(cols)
            s = math.sqrt(RSS / gl)
            
            pvals = []
            for i in range(len(cols)):
                t = float(betas[i] / (s * math.sqrt(XT_X_inv[i][i])))
                p = scipy.stats.t.sf(abs(t), df=gl) * 2
                pvals.append(p)
            
            pvals_sin_int = pvals[1:]
            if len(pvals_sin_int) == 0:
                break
            max_p = max(pvals_sin_int)
            idx_max = pvals_sin_int.index(max_p) + 1
            
            if max_p > alfa:
                incluidas.pop(idx_max)
            else:
                break
        except np.linalg.LinAlgError:
            break
    return incluidas

cols_bwd = stepwise_backward(X, Y, nombres)
nombres_bwd = [nombres[i] for i in cols_bwd]
X_bwd = X[:, cols_bwd]

XT_X_b = np.matmul(X_bwd.T, X_bwd)
XT_X_inv_b = np.linalg.inv(XT_X_b)
XT_Y_b = np.matmul(X_bwd.T, Y)
betas_b = np.matmul(XT_X_inv_b, XT_Y_b)
Y_pred_b = np.matmul(X_bwd, betas_b)
Resid_b = Y - Y_pred_b
RSS_b = float(np.matmul(Resid_b.T, Resid_b))
TSS_b = TSS
R_cuad_b = float(1 - RSS_b / TSS_b)
RSqAj_b = float(1 - (RSS_b / (X_bwd.shape[0] - X_bwd.shape[1])) / (TSS_b / (X_bwd.shape[0] - 1)))
s_cuad_b = RSS_b / (len(Y) - X_bwd.shape[1])
s_b = math.sqrt(s_cuad_b)

result_t_b = []
for i in range(0, X_bwd.shape[1]):
    t = float(betas_b[i] / (s_b * math.sqrt(XT_X_inv_b[i][i])))
    result_t_b.append(t)

gl_b = len(Y) - X_bwd.shape[1]

print(f"N° parámetros: {X_bwd.shape[1]}")
print(f"R²        = {R_cuad_b:.6f}")
print(f"R² aj.    = {RSqAj_b:.6f}")
print(f"RSS       = {RSS_b:.4e}")
print(f"s         = {s_b:.4e}")
print()
for i in range(0, X_bwd.shape[1]):
    p = scipy.stats.t.sf(abs(result_t_b[i]), df=gl_b) * 2
    sig = "SÍ" if p < 0.05 else "NO"
    print(f"Beta {i} ({nombres_bwd[i]}): t={result_t_b[i]:.4f}, p={p:.4e} → {sig} significativa")
```

```python
def stepwise_bidireccional(X, Y, nombres, alfa=0.01):
    n, k_total = X.shape
    incluidas = [0]
    candidatas = list(range(1, k_total))
    
    while True:
        cambio = False
        
        # Forward
        if candidatas:
            mejor_p = 1.0
            mejor_var = None
            for j in candidatas:
                cols = incluidas + [j]
                X_test = X[:, cols]
                try:
                    XT_X = np.matmul(X_test.T, X_test)
                    XT_X_inv = np.linalg.inv(XT_X)
                    XT_Y = np.matmul(X_test.T, Y)
                    betas = np.matmul(XT_X_inv, XT_Y)
                    Y_pred = np.matmul(X_test, betas)
                    Resid = Y - Y_pred
                    RSS = float(np.matmul(Resid.T, Resid))
                    gl = n - len(cols)
                    s = math.sqrt(RSS / gl)
                    idx = len(cols) - 1
                    t = float(betas[idx] / (s * math.sqrt(XT_X_inv[idx][idx])))
                    p = scipy.stats.t.sf(abs(t), df=gl) * 2
                    if p < mejor_p:
                        mejor_p = p
                        mejor_var = j
                except np.linalg.LinAlgError:
                    continue
            if mejor_var is not None and mejor_p < alfa:
                incluidas.append(mejor_var)
                candidatas.remove(mejor_var)
                cambio = True

        if len(incluidas) > 1:
            cols = incluidas
            X_test = X[:, cols]
            XT_X = np.matmul(X_test.T, X_test)
            XT_X_inv = np.linalg.inv(XT_X)
            XT_Y = np.matmul(X_test.T, Y)
            betas = np.matmul(XT_X_inv, XT_Y)
            Y_pred = np.matmul(X_test, betas)
            Resid = Y - Y_pred
            RSS = float(np.matmul(Resid.T, Resid))
            gl = n - len(cols)
            s = math.sqrt(RSS / gl)
            
            pvals = []
            for i in range(len(cols)):
                t = float(betas[i] / (s * math.sqrt(XT_X_inv[i][i])))
                p = scipy.stats.t.sf(abs(t), df=gl) * 2
                pvals.append(p)
            
            pvals_sin_int = pvals[1:]
            if pvals_sin_int:
                max_p = max(pvals_sin_int)
                idx_max = pvals_sin_int.index(max_p) + 1
                if max_p > alfa:
                    var_eliminada = incluidas.pop(idx_max)
                    candidatas.append(var_eliminada)
                    cambio = True
        
        if not cambio:
            break
    return incluidas

cols_bid = stepwise_bidireccional(X, Y, nombres)
nombres_bid = [nombres[i] for i in cols_bid]
X_bid = X[:, cols_bid]

XT_X_bid = np.matmul(X_bid.T, X_bid)
XT_X_inv_bid = np.linalg.inv(XT_X_bid)
XT_Y_bid = np.matmul(X_bid.T, Y)
betas_bid = np.matmul(XT_X_inv_bid, XT_Y_bid)
Y_pred_bid = np.matmul(X_bid, betas_bid)
Resid_bid = Y - Y_pred_bid
RSS_bid = float(np.matmul(Resid_bid.T, Resid_bid))
TSS_bid = TSS
R_cuad_bid = float(1 - RSS_bid / TSS_bid)
RSqAj_bid = float(1 - (RSS_bid / (X_bid.shape[0] - X_bid.shape[1])) / (TSS_bid / (X_bid.shape[0] - 1)))
s_cuad_bid = RSS_bid / (len(Y) - X_bid.shape[1])
s_bid = math.sqrt(s_cuad_bid)

result_t_bid = []
for i in range(0, X_bid.shape[1]):
    t = float(betas_bid[i] / (s_bid * math.sqrt(XT_X_inv_bid[i][i])))
    result_t_bid.append(t)

gl_bid = len(Y) - X_bid.shape[1]

print(f"N° parámetros: {X_bid.shape[1]}")
print(f"R²        = {R_cuad_bid:.6f}")
print(f"R² aj.    = {RSqAj_bid:.6f}")
print(f"RSS       = {RSS_bid:.4e}")
print(f"s         = {s_bid:.4e}")
print()
for i in range(0, X_bid.shape[1]):
    p = scipy.stats.t.sf(abs(result_t_bid[i]), df=gl_bid) * 2
    sig = "SÍ" if p < 0.05 else "NO"
    print(f"Beta {i} ({nombres_bid[i]}): t={result_t_bid[i]:.4f}, p={p:.4e} → {sig} significativa")
```
