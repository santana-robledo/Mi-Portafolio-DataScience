```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
lambda_ = 2.5
n = 1000

# Simulación
data = np.random.poisson(lam=lambda_, size=n)

# Visualización
plt.hist(data, bins=range(0, max(data)+2), density=True, alpha=0.7)
plt.title("Distribución Poisson (λ = 2.5, n = 1000)")
plt.xlabel("Número de terremotos por año")
plt.ylabel("Frecuencia relativa")
plt.grid()
plt.show()

# Estadísticas básicas
print("Media simulada:", np.mean(data))
print("Varianza simulada:", np.var(data))
```

```python
df = pd.read_csv("marketing_campaign (1).csv", sep="\t")
df.columns = df.columns.str.strip()
print(df.columns)
df["VisitGroup"] = df["NumWebVisitsMonth"].apply(
    lambda x: "High" if x > 5 else "Low"
)
```

```python
high_visits = df[df["VisitGroup"] == "High"]["NumWebPurchases"]
low_visits = df[df["VisitGroup"] == "Low"]["NumWebPurchases"]
```

```python
from scipy import stats

f_stat, p_value = stats.f_oneway(high_visits, low_visits)

print("F-statistic:", f_stat)
print("p-value:", p_value)
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.boxplot(x="VisitGroup", y="NumWebPurchases", data=df)
plt.title("Compras online vs visitas web")
plt.show()
```
