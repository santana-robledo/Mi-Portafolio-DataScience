```python
import pandas as pd
import sqlite3
import csv

conn = sqlite3.connect("RH.db")
cur = conn.cursor()
```

```python
with open("recursos_humanos.csv") as f:
    reader= csv.reader(f)
    data=list(reader)
```

```python
cur.execute("DROP TABLE IF EXISTS Detalle")

cur.execute("""
CREATE TABLE Detalle (
    satisfaction_level REAL,
    last_evaluation REAL,
    number_project INTEGER,
    average_montly_hours INTEGER,
    time_spend_company INTEGER,
    Work_accident INTEGER,
    left INTEGER,
    promotion_last_5years INTEGER,
    sales TEXT,
    salary TEXT
)
""")

cur.executemany("""
INSERT INTO Detalle VALUES (?, ?, ?, ?, ?, ?, ?, ?, ?, ?)
""", data[1:])

conn.commit()
```

```python
#Sales,Salary y satisfation level ordenados descendentemente
cur.execute("""
SELECT sales, salary, satisfaction_level
FROM Detalle
ORDER BY satisfaction_level DESC
""")

result = cur.fetchall()
result
```

```python
cur.execute("PRAGMA table_info(Detalle)")
cur.fetchall()
```

```python
#Promedio obtenido de la ultima evaluacion para cada departamento
cur.execute("""
SELECT sales, AVG(last_evaluation)
FROM Detalle
GROUP BY sales
""")

result = cur.fetchall()
result
```

```python
#Promedio de accidentes mayor a 0.15
cur.execute("""
SELECT sales, AVG(Work_accident) AS promedio_accidentes
FROM Detalle
GROUP BY sales
HAVING AVG(Work_accident) > 0.15
""")

result = cur.fetchall()
result
```

```python
cur.execute("""
SELECT sales, SUM(Work_accident) AS numero_accidentes
FROM Detalle
GROUP BY sales
HAVING SUM(Work_accident) > 200
""")

result = cur.fetchall()
result
```
