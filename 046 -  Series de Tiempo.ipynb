```python
#Prueba caminata aleatoria
#H0: Proceso de caminata aleatoria
#H1: No se tiene proceso de caminata aleatoria

from statsmodels.tsa.stattools import adfuller
import numpy as np

np.random.seed(1)
steps=np.random.normal(loc=0.001, scale=0.01, size=500)+1
```

```python
steps[0]=1
P=100*np.cumprod(steps)
```

```python
plt.plot(P)
plt.title("Proceso de caminata aleatoria")
plt.show()
```

```python
resultado=adfuller(P)
print("El valor p de la prueba aleatoria es: "+ str(resultado[1]))
```

```python
#Prueba Dickey-Fuller para los precios de Apple
resultado= adfuller(apple)
print("El valor p de la prueba aleatoria es: "+ str(resultado[1]))
```

```python
#Prediccion simple mediante promedios móviles
apple.shape
```

```python
len_train = int(apple.shape[0]*0.8)
len_train
```

```python
#Definicion de grupos de entrenamiento y prueba
train=apple[0:len_train]
test=apple[len_train:]
```

```python
len(train)
```

```python
len(test)
```

```python
y_pred= apple.copy()
y_pred=pd.DataFrame(y_pred)
```

```python
y_pred["pronostico"]= apple.rolling(30).mean()
y_pred[0:35]
```

```python
y_pred.tail()
```

```python
y_pred_m=y_pred.copy()
y_pred_m["pronostico"][29]= "Na"
for contador in range (30,752):
    y_pred_m["pronostico"][contador]=y_pred["pronostico"][contador-1]
y_pred_m[0:35]
```

```python
y_pred_m= pd.DataFrame(y_pred_m["pronostico"][604:])
y_pred_m
```

```python
plt.figure(figsize=(20,5))
plt.grid
plt.plot(train, label="Train")
plt.plot(test,label="Test")
plt.plot(y_pred_m["pronostico"],label="prediccion",color="red")
plt.legend(loc="best")
plt.title("Pronosticos mediante medias moviles")
plt.show()
```

```python
test=pd.DataFrame(test)
test
```

```python
frames= [test,y_pred_m]
result= pd.concat(frames,axis=1, join="inner")
result.columns = ["Observado", "Pronosticado"]
result
```

```python
print("Dimensiones de test:", test.shape)
print("Dimensiones de y_pred_m:", y_pred_m.shape)

print("\nÍndices de test:")
print(test.index)

print("\nÍndices de y_pred_m:")
print(y_pred_m.index)
```

```python
# Fechas que están en test, pero no en y_pred_m
fechas_faltantes = test.index.difference(y_pred_m.index)

print("Fechas faltantes en las predicciones:")
print(fechas_faltantes)
```

```python
from sklearn.metrics import mean_squared_error

# Raíz del error cuadrático medio (RMSE)
rmse = np.sqrt(mean_squared_error(test.iloc[:148], y_pred_m)).round(2)

# Error porcentual absoluto medio (MAPE)
mape = (np.mean(np.abs(test.iloc[:148, 0] - y_pred_m["pronostico"]) / test.iloc[:148, 0]) * 100).round(2)

print("RMSE:", rmse)
print("MAPE:", mape, "%")
    
```
