```python
import yfinance as yf
import pandas as pd
import warnings
warnings.filterwarnings("ignore")
import matplotlib.pyplot as plt
```

```python
apple = yf.download(tickers='AAPL', period='5y',interval='1mo',rounding=True)
apple
```

```python
apple = apple["Close"]
apple
```

```python
apple.index = pd.to_datetime(apple.index)
apple2022 = apple[apple.index.year == 2022]
apple2022
```

```python
apple2022.plot(grid=True)
```

```python
rendimientos = apple.pct_change()
rendimientos
```

```python
aaple_microsoft = yf.download(tickers='AAPL MSFT', period='5y',interval='1mo',rounding=True)
aaple_microsoft
```

```python
aaple_microsoft=aaple_microsoft["Close"]
aaple_microsoft
```

```python
aaple_microsoft["AAPL"].plot()
aaple_microsoft["MSFT"].plot()
plt.legend(["Apple","Microsoft"])
plt.title("Apple Microsoft")
plt.show()
```

```python
rend_apple_microsoft = aaple_microsoft.pct_change()
rend_apple_microsoft=rend_apple_microsoft.dropna()
rend_apple_microsoft
```

```python
correlacion = rend_apple_microsoft["AAPL"].corr(rend_apple_microsoft["MSFT"])
print("Correlacion entre Apple y Microsoft",correlacion)
```

```python
plt.scatter(rend_apple_microsoft["AAPL"],rend_apple_microsoft["MSFT"])
plt.show()
```

```python
import statsmodels.api as sm
rend_apple_microsoft = sm.add_constant(rend_apple_microsoft)
rend_apple_microsoft
```
