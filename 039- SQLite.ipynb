```python
flavor = input("Introduzca un sabor: ")
rating = input("Introduzca una calificación: ")

cur.execute("INSERT INTO icecream_flavors (flavor, rating) VALUES (?, ?)",(flavor, rating))
conn.commit()
```

```python
flavor = input("Introduzca un sabor: ")
rating = input("Introduzca una calificación: ")

cur.execute("INSERT INTO icecream_flavors (flavor, rating) VALUES (:sabor, :puntaje)",{"sabor":flavor, "puntaje":rating})
conn.commit()
```

```python
ice_cream_flavors = [("Vainilla",4.5),("Fresa",9),("Cafe",6),("Caramelo",7.5)]
```

```python
cur.executemany(f"INSERT INTO icecream_flavors(flavor,rating) VALUES (?,?);", ice_cream_flavors)
conn.commit()
```

```python
cur.executescript("""
CREATE TABLE IF NOT EXISTS toppings(
    id INTEGER PRIMARY KEY,
    name TEXT NOT NULL
);

INSERT INTO toppings(name)
VALUES
    ("sprinkles"),
    ("chocosauce"),
    ("caramelosauce"),
    ("nuts");
""")

conn.commit()
```

```python
for line in conn.iterdump():
    print(line)
```

```python
_file = open("ice_cream_db_dump.sql", "w")

for line in conn.iterdump():
    _file.write(f"{line}\n")

_file.close()
```
