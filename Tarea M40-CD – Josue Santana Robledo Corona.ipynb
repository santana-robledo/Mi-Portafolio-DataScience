```python
import sqlite3
import csv

conn = sqlite3.connect("Base.db")
cur = conn.cursor()
```

```python
with open("Customers.csv", encoding="utf-8") as f:
    reader = csv.DictReader(f)
    customers = list(reader)
```

```python
clientes_seleccionados = customers[:5]

clientes_seleccionados
```

```python
cur.execute("""
CREATE TABLE IF NOT EXISTS Clientes (
    CustomerID INTEGER PRIMARY KEY,
    CustomerName TEXT NOT NULL,
    ContactName TEXT,
    Address TEXT,
    City TEXT,
    PostalCode TEXT,
    Country TEXT,
    PurchaseUSD REAL
);
""")

conn.commit()
```

```python
for cliente in clientes_seleccionados:
    cur.execute("""
        INSERT INTO Clientes (
            CustomerID,
            CustomerName,
            ContactName,
            Address,
            City,
            PostalCode,
            Country,
            PurchaseUSD
        )
        VALUES (?, ?, ?, ?, ?, ?, ?, ?);
    """, (
        cliente["CustomerID"],
        cliente["CustomerName"],
        cliente["ContactName"],
        cliente["Address"],
        cliente["City"],
        cliente["PostalCode"],
        cliente["Country"],
        cliente["PurchaseUSD"]
    ))

conn.commit()
```

```python
cur.execute("SELECT * FROM Clientes;")
result = cur.fetchall()
result
```

```python
with open("Operaciones.sql", "w", encoding="utf-8") as f:
    for line in conn.iterdump():
        f.write(line + "\n")
```
