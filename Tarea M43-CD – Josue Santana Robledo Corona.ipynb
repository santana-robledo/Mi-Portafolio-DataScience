# Inner Join Ejemplo 1
Para esta consulta se utiliza INNERJOIN mediante el campo CustomerName, el resultado contiene unicamente  los clientesque tienen coincidencia en ambas tablas

```python
cur.execute("""
SELECT 
    Customers1.CustomerName,
    Customers1.City,
    Customers1.Country,
    Customers2.PurchaseUSD
FROM Customers1
INNER JOIN Customers2
    ON Customers1.CustomerName = Customers2.CustomerName;
""")

result = cur.fetchall()
result
```

# Inner Join Ejemplo 2
Para esta consulta Se obtiene información de ubicación desde Customers1 y datos de contacto desde Customers2.

```python
cur.execute("""
SELECT 
    Customers1.CustomerName,
    Customers1.City,
    Customers1.Country,
    Customers2.ContactName,
    Customers2.PostalCode
FROM Customers1
INNER JOIN Customers2
    ON Customers1.CustomerName = Customers2.CustomerName;
""")

result = cur.fetchall()
result
```

# LEFT JOIN Ejemplo 1
En este caso se utiliza LEFT JOIN, por lo que se conservan todos los clientes de Customers1, independientemente de que exista o no información correspondiente en Customers2

```python

cur.execute("""
SELECT 
    Customers1.CustomerName,
    Customers1.City,
    Customers1.Country,
    Customers2.PurchaseUSD
FROM Customers1
LEFT JOIN Customers2
    ON Customers1.CustomerName = Customers2.CustomerName;
""")

result = cur.fetchall()
result
```

# LEFT JOIN Ejemplo 2
Esta consulta utiliza LEFT JOIN para conservar todos los clientes de Customers1 y posteriormente utiliza IS NULL para identificar aquellos que no tienen una coincidencia en Customers2

```python

cur.execute("""
SELECT 
    Customers1.CustomerName,
    Customers1.City,
    Customers1.Country
FROM Customers1
LEFT JOIN Customers2
    ON Customers1.CustomerName = Customers2.CustomerName
WHERE Customers2.CustomerName IS NULL;
""")

result = cur.fetchall()
result
```

# CASE WHEN EJEMPLO 1
La instrucción CASE WHEN permite clasificar los clientes de acuerdo con el monto de sus compras.

```python
#CASE WHEN EJEMPLO 1
cur.execute("""
SELECT 
    CustomerName,
    PurchaseUSD,
    CASE
        WHEN PurchaseUSD < 500 THEN 'Compra baja'
        WHEN PurchaseUSD < 1000 THEN 'Compra media'
        ELSE 'Compra alta'
    END AS CategoriaCompra
FROM Customers2;
""")

result = cur.fetchall()
result
```

# Ejemplo Subquery

La consulta utiliza una SubQuery para calcular primero el promedio de PurchaseUSD de todos los clientes. Posteriormente, seleccionamos únicamente aquellos clientes cuyo monto de compra es superior al promedio.

```python
#Ejemplo SUBQuery
cur.execute("""
SELECT 
    CustomerName,
    PurchaseUSD
FROM Customers2
WHERE PurchaseUSD > (
    SELECT AVG(PurchaseUSD)
    FROM Customers2
);
""")

result = cur.fetchall()
result
```
