```python
#Lectura del archivo CSV
import os
import csv

os.chdir(r"C:\Users\Josue Santana Robled")

with open("Customers.csv") as f:
    reader = csv.reader(f)
    data = list(reader)
```

```python
data
```

```python
len(data)
```

```python
#Creación de la tabla donde se importará el archivo CSV
cur.execute("""
CREATE TABLE customers_2(
CustomerID REAL PRIMARY KEY,
CustomerName TEXT,
ContactName TEXT,
Adress TEXT,
City TEXT,
PostalCode TEXT,
Country TEXT,
PurchaseUSD REAL
);
""")
```

```python
#Inserción de datos en la tabla creada
for renglon in range(1, len(data)):
    cur.execute("""
        INSERT INTO customers_2(CustomerID,CustomerName,ContactName,Adress,City,PostalCode,Country,PurchaseUSD)
        VALUES(?,?,?,?,?,?,?,?)
    """,data[renglon])
```

```python
conn.commit()
```

```python
conn.close()
```
