```python
import pandas as pd
import numpy as np
import warnings
warnings.filterwarnings("ignore")
import os

```

```python
data_vw= pd.read_csv("vw.csv")
data_vw
```

```python
data_vw["Intercepto"]=1
data_vw= data_vw[["Intercepto","year", "mileage","tax","mpg","engineSize","price"]]
print(data_vw.shape)
data_vw.head()
```

```python
Xdata = data_vw[["Intercepto","year","mileage","tax","mpg","engineSize"]].values
Ydata = data_vw[["price"]].values
```

```python
#Dividir en bases de entrenamiento y pruebas
from sklearn.model_selection import train_test_split
X_train,X_test,Y_train,Y_test= train_test_split(Xdata,Ydata,test_size=0.3,random_state=1)
```

```python
X=X_train
Y=Y_train
```

```python
#Numero de renglones en Matriz X
X.shape[0]
```

```python
#Numero de columnas en Matriz X
X.shape[1]
```

```python
np.set_printoptions(formatter={"float_kind": "{:f}".format})
```

```python
XT_X=np.matmul(np.matrix.transpose(X),X)
XT_X
```

```python
XT_X_inv=np.linalg.inv(XT_X)
XT_X_inv
```

```python
XT_Y=np.matmul(np.matrix.transpose(X),Y)
XT_Y
```

```python
betas=np.matmul(XT_X_inv,XT_Y)
betas
```

```python
#Calculo de TSS (Suma del Total de cuadrados)
TSS= np.matmul(np.matrix.transpose(Y),Y)
TSS
```

```python
#Calculo de ESS (Fórmula explicada por el módelo)
ESS= np.matmul(np.matmul(np.matrix.transpose(betas),np.matrix.transpose(X)), np.matmul(X,betas)) - len(Y)*(Y.mean()**2)
ESS

```

```python
# RSS (Cálculo de residuales al Cuadrado)
RSS= TSS - ESS
RSS
```

```python
#Cáculo de coeficiente de Determinación R cuadrada
RSq = 1- RSS/TSS
RSq
```

```python
#Cáculo del Coeficiente de determinación Múltiple
RSqAj = 1- (RSS/(X.shape[0]-X.shape[1])) / (TSS / X.shape[0]-1)
RSqAj
```

```python
import statsmodels.api as sm

regressor= sm.OLS(Y,X).fit()
print(regressor.summary())
```

```python
#Aplicacion sobre Base de Prueba
Y_pred= np.matmul(X_test,betas)
Y_pred
```

```python
#Calculo de residuales
Resid = Y_test - Y_pred
Resid
```

```python
#Grafico del histograma de residuales para la base de prueba
df = pd.DataFrame(Resid)
df.hist()
```
